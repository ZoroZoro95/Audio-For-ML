```python
import numpy as np
import matplotlib.pyplot as plt

sample_rate = 1000
duration = 1.0

number_of_samples = int(sample_rate * duration)
time = np.arange(number_of_samples) / sample_rate

wave_1 = np.sin(2 * np.pi * 5 * time)
wave_2 = 0.6 * np.sin(2 * np.pi * 12 * time)
wave_3 = 0.3 * np.sin(2 * np.pi * 25 * time)

combined_signal = wave_1 + wave_2 + wave_3
```

```python
def calculate_similarity(signal, frequency, time):
    test_wave = np.sin(
        2 * np.pi * frequency * time
    )

    multiplied = signal * test_wave

    score = np.sum(multiplied)

    return score
```

```python
for frequency in [5, 8, 12, 18, 25]:
    score = calculate_similarity(
        combined_signal,
        frequency,
        time
    )

    print(f"{frequency} Hz: {score}")
```
