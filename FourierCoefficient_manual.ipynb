```python
import numpy as np 

sample_rate = 1000
duration = 1.0 #1sec
N = sample_rate * duration
t = np.arange(N) / sample_rate

#signal with phase pi/4 and 5hz freq
signal = np.sin(
    2 * np.pi * 5 * t + np.pi/4 
)

test_freq = 5 #5hz test freq

```

```python
#complex test wave is e^(-2jpift)

complex_test_wave = np.exp(-2j * np.pi * test_freq * t)
```

```python
coefficient = np.sum(signal * complex_test_wave)
```

```python
print("Complex coefficient:", coefficient) 
print("Real part:", coefficient.real) 
print("Imaginary part:", coefficient.imag) 
print("Magnitude:", np.abs(coefficient)) 
print("Phase:", np.angle(coefficient))
```
