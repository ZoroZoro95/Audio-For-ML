```python
import numpy as np

#manual dft for only one frequency bin k = 1
signal = np.array([1,0,-1,0],dtype = float)

#no of samples
N = len(signal)
k = 1

#initialise the complex coefficient
coefficient = 0j

#n is index for time domain

for n in range(N):
    angle = (-2j * (np.pi * k * n))/N
    test_wave_value = np.exp(angle)

    product = signal[n] * test_wave_value

    coefficient += product

    print(
        f"n={n}, "
        f"x[n]={signal[n]}, "
        f"test={test_wave_value}, "
        f"product={product}"
    )
print("X[1] = :", coefficient)
    
```
