```python
import numpy as np 
from scipy.fft import dct

signal = np.array([1,2,3,4,5,6,7,8],dtype=float)

coefficients = dct(signal , type = 2, norm = "ortho")
print(coefficients)
```

```python
"""
9. What does norm="ortho" mean?
norm="ortho"
applies an orthonormal scaling.
It makes the transform mathematically cleaner and easier to invert.
Then you can reconstruct the signal with the inverse DCT:
"""
from scipy.fft import idct

reconstructed = idct(
    coefficients,
    type=2,
    norm="ortho"
)

print(reconstructed)
```

```python
"""
10. DCT coefficients
Just like the DFT gives:

X[0],X[1],...
the DCT gives: C[0],C[1],C[2],..
But these coefficients are real numbers.
Conceptually:
C[0]: very low-frequency / average-like component
C[1]: slow variation
C[2]: slightly faster variation
higher k: faster oscillations
So lower DCT indices correspond to smoother structure.
"""
```

```python
#smooth/gradual signal
import numpy as np
from scipy.fft import dct

signal = np.array([
    1.0,
    1.1,
    1.2,
    1.3,
    1.4,
    1.5,
    1.6,
    1.7
])

coefficients = dct(
    signal,
    type=2,
    norm="ortho"
)

print(coefficients)
```

```python
"""
You will likely see:
large values in the first few coefficients
small values in higher coefficients
That is energy compaction.
"""
```

```python
"""
n a DCT used on something like a mel spectrum, the DCT coefficient index is not simply:
coefficient 5 = 5 Hz
It describes how rapidly values vary across the input dimension.
So don't interpret DCT coefficients exactly like FFT bins.
"""
```

```python
import numpy as np
from scipy.fft import dct, idct

signal = np.array([
    1.0,
    1.1,
    1.2,
    1.3,
    1.4,
    1.5,
    1.6,
    1.7
])

coefficients = dct(
    signal,
    type=2,
    norm="ortho"
)

compressed = coefficients.copy()

compressed[3:] = 0

reconstructed = idct(
    compressed,
    type=2,
    norm="ortho"
)

print("Original:")
print(signal)

print("DCT:")
print(coefficients)

print("Reconstructed:")
print(reconstructed)
```

```python
#reconstrucuted and original are almost similar even though we set all the coefficients from 3 to 8 = 0
#this is how compression works
```
