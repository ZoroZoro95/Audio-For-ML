```python
import librosa
```

```python
frequencies_hz = [0, 100, 500, 1000, 2000, 4000, 8000]

mel_values = librosa.hz_to_mel(frequencies_hz, htk = True)

print(mel_values)
```

```python
"""
In this context, htk=True means Librosa uses the HTK version of the Mel scale, which follows the classic formula:

m=2595log10(1 + f/700)
"""
```

```python
hz_values = librosa.mel_to_hz(mel_values, htk= True)

print(hz_values)
```

```python
import numpy as np

# --------------------------------------------------
# Step 1: Power spectrum from 5 FFT bins
# --------------------------------------------------

power_spectrum = np.array([
    2.0,
    4.0,
    8.0,
    6.0,
    2.0
])

print("Power spectrum:")
print(power_spectrum)

# --------------------------------------------------
# Step 2: Manually define 2 Mel filters
# --------------------------------------------------

mel_filter_1 = np.array([
    0.0,
    0.5,
    1.0,
    0.5,
    0.0
])

mel_filter_2 = np.array([
    0.0,
    0.0,
    0.5,
    1.0,
    0.5
])

# Stack them into one filterbank matrix
mel_filterbank = np.array([
    mel_filter_1,
    mel_filter_2
])

print("\nMel filterbank shape:")
print(mel_filterbank.shape)

print("\nMel filterbank:")
print(mel_filterbank)

# --------------------------------------------------
# Step 3: Calculate each Mel-band energy manually
# --------------------------------------------------

mel_band_1 = np.sum(
    power_spectrum * mel_filter_1
)

mel_band_2 = np.sum(
    power_spectrum * mel_filter_2
)

print("\nMel band 1 energy:")
print(mel_band_1)

print("\nMel band 2 energy:")
print(mel_band_2)

# --------------------------------------------------
# Step 4: Store final Mel representation
# --------------------------------------------------

mel_energies_manual = np.array([
    mel_band_1,
    mel_band_2
])

print("\nManual Mel energies:")
print(mel_energies_manual)

# --------------------------------------------------
# Step 5: Same calculation using matrix multiplication
# --------------------------------------------------

mel_energies_matrix = (
    mel_filterbank @ power_spectrum
)

print("\nUsing matrix multiplication:")
print(mel_energies_matrix)

print(
    "\nSame result:",
    np.allclose(
        mel_energies_manual,
        mel_energies_matrix
    )
)
```
