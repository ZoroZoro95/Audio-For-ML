```python
import numpy as np

#manual dft for only one frequency bin k = 1
signal = np.array([1,0,-1,0],dtype = float)

#no of samples
N = len(signal)
k = 1

#initialise the complex coefficient
coefficient = 0j

#n is index for time domain

for n in range(N):
    angle = (-2j * (np.pi * k * n))/N
    test_wave_value = np.exp(angle)

    product = signal[n] * test_wave_value

    coefficient += product

    print(
        f"n={n}, "
        f"x[n]={signal[n]}, "
        f"test={test_wave_value}, "
        f"product={product}"
    )
print("X[1] = :", coefficient)
    
```

```python
def calculate_dft(signal):
    signal = np.asarray(signal,dtype=float)
    N = len(signal)
    spectrum = np.zeros(N,dtype=complex) # creates [0j,0j,0j,0j]
    for k in range(N):
        coefficient = 0j
        for n in range(N):
            angle = -2j * np.pi * k * n / N

            test_wave_value = np.exp(angle)
        
            product = signal[n] * test_wave_value
        
            coefficient += product
        spectrum[k] = coefficient
    return spectrum
```

```python
signal = np.array([1, 0, -1, 0],dtype = float)

dft_result = calculate_dft(signal)

print(dft_result)
#the e^-16 is floating point error when calculating sin and cos values
#[0.+0.j 2.+0.j 0.+0.j 2.+0.j] real answer we can round off in future if reqd
```

```python
#comparison with numpy fft
numpy_result = np.fft.fft(signal)

print("Manual DFT:")
print(dft_result)

print("\nNumPy FFT:")
print(numpy_result)

print("\nSame result:")
print(np.allclose(dft_result, numpy_result))
```
