```python
import numpy as np

sample_rate = 1000
duration = 1.0

number_of_samples = int(sample_rate * duration)
time = np.arange(number_of_samples) / sample_rate

# 5 Hz frequency, but shifted by pi/2
signal = np.sin(
    2 * np.pi * 5 * time + np.pi / 2
)

sine_test = np.sin(
    2 * np.pi * 5 * time
)

cosine_test = np.cos(
    2 * np.pi * 5 * time
)

sine_score = np.sum(signal * sine_test)
cosine_score = np.sum(signal * cosine_test)

print("Sine score:", sine_score)
print("Cosine score:", cosine_score)
```

```python
#magnitude needs to take into consideration both sine and cosine test results
import numpy as np

sample_rate = 1000
duration = 1.0

number_of_samples = int(sample_rate * duration)
time = np.arange(number_of_samples) / sample_rate

frequency = 5

sine_test = np.sin(
    2 * np.pi * frequency * time
)

cosine_test = np.cos(
    2 * np.pi * frequency * time
)

phases = [
    0,
    np.pi / 4,
    np.pi / 2,
    3 * np.pi / 4
]

for phase in phases:
    signal = np.sin(
        2 * np.pi * frequency * time + phase
    )

    sine_score = np.sum(signal * sine_test)
    cosine_score = np.sum(signal * cosine_test)

    magnitude = np.sqrt(
        sine_score**2 + cosine_score**2
    )

    print(f"Phase: {phase:.2f}")
    print(f"Sine score: {sine_score:.2f}")
    print(f"Cosine score: {cosine_score:.2f}")
    print(f"Magnitude: {magnitude:.2f}")
    print()
```

```python
#frequency bin
"""
A frequency bin is one discrete frequency position checked by the DFT/FFT.

For an N-point FFT with sample rate f_s
    f_k = (k * f_s)/N

where k is the bin index.

Example:

f_s=1000 Hz sampling freq ,N=1000 is no of samples

Then bin spacing is:

Δf= f_s/N =1 Hz


So:

X[0]→0 Hz
X[1]→1 Hz
X[5]→5 Hz
X[12]→12 Hz

Here, X[5] is the Fourier coefficient stored in the 5 Hz frequency bin. Its magnitude tells us how strongly 5 Hz is present, and its angle tells us the phase.
"""
```

```python
#fourier coefficient
"""
suppose X[5] = 300 - 400j
means fourier coefficient corresponding to bin(index) 5 and 5hz freq is 300 - 400j the complete complex value
we use this coefficient to obtain for freq which is at bin 5 its Phase and magnitude

mag = sqrt(img(X[5)^2 + Re(X[5)^2)
phase = atan2(Im(X[k]),Re(X[k])
"""
```

```python
import numpy as np

fourier_coefficient = 300 - 400j

phase_1 = np.angle(fourier_coefficient)

phase_2 = np.arctan2(
    fourier_coefficient.imag,
    fourier_coefficient.real
)

print(phase_1)
print(phase_2)
```
