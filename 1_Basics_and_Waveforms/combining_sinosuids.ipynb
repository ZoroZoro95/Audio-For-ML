```python
#we would want to represent a complex wave as combination of sinosuids
import numpy as np
import matplotlib.pyplot as plt
```

```python
sample_rate = 1000
duration = 1.0
no_of_samples = sample_rate * duration
t = np.arange(no_of_samples)/sample_rate

x1 = np.sin(2 * np.pi * 5 * t)
x2 = np.sin(2* np.pi * 12 * t)
x3 = np.sin(2 * np.pi * 25 * t)

combined_signal = x1 + x2 + x3

```

```python
plt.figure(figsize=(12, 8))

plt.subplot(4, 1, 1)
plt.plot(t, x1)
plt.title("5 Hz, amplitude 1.0")
plt.ylabel("Amplitude")
plt.xlim(0, 0.5)
plt.grid()

plt.subplot(4, 1, 2)
plt.plot(t, x2)
plt.title("12 Hz, amplitude 0.6")
plt.ylabel("Amplitude")
plt.xlim(0, 0.5)
plt.grid()

plt.subplot(4, 1, 3)
plt.plot(t, x3)
plt.title("25 Hz, amplitude 0.3")
plt.ylabel("Amplitude")
plt.xlim(0, 0.5)
plt.grid()

plt.subplot(4, 1, 4)
plt.plot(t, combined_signal)
plt.title("Combined signal")
plt.xlabel("Time in seconds")
plt.ylabel("Amplitude")
plt.xlim(0, 0.5)
plt.grid()

plt.tight_layout()
plt.show()
```

```python
#checking amp values for index 100
index = 100

print("Time:", t[index])
print("Wave 1:", x1[index])
print("Wave 2:", x2[index])
print("Wave 3:", x3[index])
print("Combined:", combined_signal[index])
```
