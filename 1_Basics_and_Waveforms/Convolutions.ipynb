```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# Input signal
# --------------------------------------------------

signal = np.array([
    1.0,
    2.0,
    1.0
])

# --------------------------------------------------
# Simple filter
# --------------------------------------------------

filter_kernel = np.array([
    1.0,
    0.5
])

# --------------------------------------------------
# Convolution
# --------------------------------------------------

output = np.convolve(
    signal,
    filter_kernel,
    mode="full"
)

print("Input:")
print(signal)

print("\nFilter:")
print(filter_kernel)

print("\nOutput:")
print(output)
```

```python
import numpy as np
import matplotlib.pyplot as plt

sample_rate = 16000
duration = 0.5
frequency = 440

N = int(
    sample_rate * duration
)

time = (
    np.arange(N)
    / sample_rate
)

signal = np.sin(
    2 * np.pi * frequency * time
)

# --------------------------------------------------
# Build echo filter
# --------------------------------------------------

echo_delay = 0.1

delay_samples = int(
    echo_delay * sample_rate
)

impulse_response = np.zeros(
    delay_samples + 1
)

# Original sound
impulse_response[0] = 1.0

# Echo 100 ms later
impulse_response[delay_samples] = 0.5

# --------------------------------------------------
# Apply convolution
# --------------------------------------------------

output = np.convolve(
    signal,
    impulse_response
)

# --------------------------------------------------
# Time axes
# --------------------------------------------------

output_time = (
    np.arange(len(output))
    / sample_rate
)

# --------------------------------------------------
# Plot input
# --------------------------------------------------

plt.figure(figsize=(12, 4))

plt.plot(
    time,
    signal
)

plt.title("Original Signal")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")

plt.show()

# --------------------------------------------------
# Plot output
# --------------------------------------------------

plt.figure(figsize=(12, 4))

plt.plot(
    output_time,
    output
)

plt.title("Signal After Echo Convolution")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")

plt.show()
```
