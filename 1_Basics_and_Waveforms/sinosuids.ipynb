```python
import numpy as np
import matplotlib.pyplot as plt

sample_rate = 1000
duration = 1.0

number_of_samples = int(sample_rate * duration)

time = np.arange(number_of_samples) / sample_rate

amplitude = 1.0
frequency = 5
phase = np.pi

signal = amplitude * np.sin(
    2 * np.pi * frequency * time + phase
)

plt.figure(figsize=(12, 4))
plt.plot(time, signal)

plt.xlabel("Time in seconds")
plt.ylabel("Amplitude")
plt.title("5 Hz sine wave")
plt.grid()

plt.show()
```
